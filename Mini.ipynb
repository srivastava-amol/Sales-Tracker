```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
pd.set_option('display.float_format', lambda x: '%.2f' % x)

df = pd.read_csv('index_1.csv')
df1 = pd.read_csv('index_2.csv')

df['time'] = pd.to_datetime(df['datetime']).dt.time
df = df.drop(columns=['datetime'], errors='ignore')

print("--- Dataset Info ---")
print(df.info())

print("\n--- First 5 Rows ---")
print(df.head().to_string())

df['date'] = pd.to_datetime(df['date']).dt.round('s')
df['DayOfWeek'] = df['date'].dt.day_name()

print("\n--- Missing Values ---")
print(df.isnull().sum())
df = df.dropna()

print("\n--- Summary Statistics ---")
print(df.describe(include='number'))

daily_sales = df.groupby('date')['money'].sum().reset_index()
category_sales = df.groupby('coffee_name')['money'].sum().reset_index()

sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 5))
plt.plot(daily_sales['date'], daily_sales['money'], marker='o', color='b', linestyle='-')
plt.title('Daily Total Sales Trend')
plt.xlabel('Date')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 5))
sns.barplot(
    x='coffee_name',
    y='money',
    data=category_sales,
    palette='viridis',
    hue='coffee_name',
    legend=False
)
plt.title('Total Sales by Coffee Type')
plt.xlabel('Coffee Name')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 5))
sns.boxplot(
    x='coffee_name',
    y='money',
    data=df,
    palette='Set2',
    hue='coffee_name',
    legend=False
)
plt.title('Sales Distribution by Coffee Type')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
